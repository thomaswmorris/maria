```python
import maria
import numpy as np
from maria import Instrument

xi = np.random.uniform(size=600, low=-20, high=20)
eta = np.random.uniform(size=600, low=-20, high=20)
gamma = np.random.uniform(size=600, low=0, high=2 * np.pi)

array = {"xi": xi, 
         "eta": eta,
         "gamma": gamma,
         "primary_size": 0.5,
         "bands": [maria.get_band("act/pa5/f090"),
                   maria.get_band("act/pa5/f150")]}

instrument = Instrument(arrays=[array])

instrument.plot()
```
